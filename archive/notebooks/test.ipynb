```python
import sys
import subprocess
from pathlib import Path

runner = Path(
    "/scratch/project_2009497/Old_Pipelines_Vaibhav/Old_Splits/"
    "Fidesz1_Split_Test/hungary_fidesz1_test_32/run_hungary_splits.py"
)

def launch(check_only=False):
    command = [
        sys.executable, "-u", str(runner),
        "--workers", "32",
        "--scope", "Fidesz/1",
    ]

    if check_only:
        command.append("--check-only")

    with subprocess.Popen(
        command,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    ) as process:
        for line in process.stdout:
            print(line, end="", flush=True)

        returncode = process.wait()

    if returncode:
        raise RuntimeError("The command failed. See the error printed above.")

launch(check_only=True)
```

```python
import os
import sys
import subprocess
from pathlib import Path

runner = Path(
    "/scratch/project_2009497/Old_Pipelines_Vaibhav/Old_Splits/"
    "Fidesz1_Split_Test/hungary_fidesz1_test_32/run_hungary_splits.py"
)
assert runner.is_file(), f"Launcher not found: {runner}"

# Restore notebook-compatible packages.
# MoviePy 2.2.1 was installed before the previous downgrade.
subprocess.check_call([
    sys.executable, "-m", "pip", "install",
    "--user", "--force-reinstall", "--no-deps",
    "moviepy==2.2.1",
    "decorator==5.2.1",
])

# Keep the splitter's environment separate from Jupyter.
split_env = os.environ.copy()
for name in (
    "PYTHONPATH", "PYTHONHOME",
    "PIP_USER", "PIP_TARGET", "PIP_PREFIX",
):
    split_env.pop(name, None)

split_env["PYTHONNOUSERSITE"] = "1"

runtime_folder = runner.parent / "splitter_runtime_v1"

subprocess.check_call([
    sys.executable, "-m", "venv", str(runtime_folder)
], env=split_env)

split_python = runtime_folder / "bin" / "python"

# Install the versions tested with your unchanged processors.
subprocess.check_call([
    str(split_python), "-m", "pip", "install",
    "moviepy==1.0.3",
    "decorator==4.4.2",
    "ffmpeg-python==0.2.0",
    "numpy==1.26.4",
    "opencv-python-headless==4.11.0.86",
], env=split_env)

subprocess.check_call([
    str(split_python), "-m", "pip", "check"
], env=split_env)
```

```python
def launch(check_only=False):
    command = [
        str(split_python), "-u", str(runner),
        "--workers", "32",
        "--scope", "Fidesz/1",
    ]

    if check_only:
        command.append("--check-only")

    with subprocess.Popen(
        command,
        env=split_env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
    ) as process:
        for line in process.stdout:
            print(line, end="", flush=True)

        returncode = process.wait()

    if returncode:
        raise RuntimeError("The command failed. See the output above.")

launch(check_only=True)
```
